A. WORD ROTATION PROBLEM

```python
def rotate_word(text): #This slices from index 1 to the end, then it appends the index 0.
 return text[1:] + text[0]
```

```python
#Code testing
print(rotate_word("python"))
print(rotate_word("logic"))
print(rotate_word("Code"))
print(rotate_word("A"))
```

B. USERNAME BUILDER PROBLEM

```python
def make_username(first_name, last_name):
  #Convert first name and last name to lowercases and remove the spaces.
  #Combine the first name and the last name with a period.
  return first_name.lower().replace(" ", "") + "." + last_name.lower().replace(" ","")
```

```python
#Code testing
print(make_username("Ada", "Lovelace"))
print(make_username("Alan", "Turing"))
print(make_username("Ana Maria", "De Leon"))
```

C.BOOKEND SWAP PROBLEM

```python
def swap_bookends(items):
  #Make the first, *middle, and last elements as a list.
 first, *middle, last = items
 #Then we combine the list with last element first and the first element to last.
 return [last] + middle + [first]
```

```python
#Code testing
print(swap_bookends([1, 2, 3, 4, 5, 6]))
print(swap_bookends(["red", "green", "blue"]))
print(swap_bookends([8, 3]))
```
